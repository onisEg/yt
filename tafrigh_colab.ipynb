# 🎙️ تفريغ الفيديوهات — Tafrigh

حط لينك أي فيديو (يوتيوب / تيك توك / فيسبوك / إنستجرام / تويتر…) واستلم كلامه نص.

**طريقة الاستخدام:**
1. من القائمة فوق: `Runtime` ← `Change runtime type` ← اختار **T4 GPU** (اختياري بس بيخلّيها أسرع بمراحل).
2. شغّل الخانة الأولى (زرار ▶) — دي التجهيز، مرة واحدة بس.
3. حط اللينك في الخانة التانية وشغّلها.

الأداة بتدوّر الأول على الترجمة الجاهزة بتاعة الفيديو (ثواني)، ولو ملقتش بتفرّغ الصوت بـ Whisper محليًا هنا في الجلسة.

```python
#@title ①  التجهيز — شغّله مرة واحدة { display-mode: "form" }
!pip install -q -U yt-dlp faster-whisper
!wget -q -O tafrigh.py https://raw.githubusercontent.com/onisEg/yt/main/tafrigh.py
print("✅ جاهز — كمّل على الخانة اللي تحت")
```

```python
#@title ②  حط اللينك ودوس ▶ { display-mode: "form" }
video_url = ""  #@param {type:"string"}
lang = "auto"  #@param ["auto", "ar", "en", "es", "tr", "fr", "de"]
model_size = "small"  #@param ["tiny", "base", "small", "medium", "large-v3"]
force_whisper = False  #@param {type:"boolean"}
make_srt = False  #@param {type:"boolean"}
auto_download = True  #@param {type:"boolean"}

import subprocess, glob, os
from IPython.display import display, Markdown

assert video_url.strip(), "❌ حط لينك الفيديو الأول"

cmd = ["python", "tafrigh.py", video_url.strip(), "-m", model_size, "-o", "/content/transcripts"]
if lang != "auto":
    cmd += ["-l", lang]
if force_whisper:
    cmd += ["-w"]
if make_srt:
    cmd += ["--srt"]

subprocess.run(cmd)

files = sorted(glob.glob("/content/transcripts/*.txt"), key=os.path.getmtime)
if files:
    newest = files[-1]
    text = open(newest, encoding="utf-8").read()
    display(Markdown(f"### 📄 {os.path.basename(newest)}\n\n---\n\n{text}"))
    if auto_download:
        from google.colab import files as colab_files
        colab_files.download(newest)
        for extra in glob.glob("/content/transcripts/*.srt") + glob.glob("/content/transcripts/*.vtt"):
            colab_files.download(extra)
else:
    print("مطلعش ولا ملف — راجع اللينك أو جرّب force_whisper.")
```
